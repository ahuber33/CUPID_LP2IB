```python
import numpy as np

def write_file(filename, energy, val, unit_energy, unit_val):
    path = r"/home/local1/Documents/PYTHON//"
    
    if unit_val=="":
        f = open(path+filename+"_"+str(unit_energy)+".txt", "x")
    else:
        f = open(path+filename+"_"+str(unit_energy)+"_"+str(unit_val)+".txt", "x")
        
    for i in range(len(energy)):
        if i != len(energy)-1:
            f.write(str(energy[i]) + ", " + str(val[i]) + "\n")
        else:
            f.write(str(energy[i]) + ", " + str(val[i]))

        
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

df = pd.read_csv("LMO_ABSORPTION_eV_mm.txt", header=None)
display(df)
energy = df[0].values
var = df[1].values
var2 = np.exp(-50/var)*100
wavelength = 1240/energy

def eV_to_nm(eV):
    return 1240/eV

def nm_to_eV(nm):
    return 1240/nm

fig, ax1 = plt.subplots(layout='constrained', figsize=(4,3))
ax1.plot(wavelength, var2, color='black')
ax2 = ax1.secondary_xaxis('top', functions=(eV_to_nm,nm_to_eV))
ax1.set_xlabel('wavelength [nm]')
ax2.set_xlabel('energy [eV]')
ax1.set_ylabel('transmittance [%]')
fig.savefig('transmittance',dpi=300)
```
